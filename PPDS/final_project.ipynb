 In later cells I want to calculate the gross in each year, so I am converting the release date to a datetime     object and extracting the year from it. I am keeping that extracted year in a separate column in the same data  frame

```python
# extract the year from release_date and set it as another column
Disney_total_gross_with_year_column = (Disney_total_gross
    .assign(year=  pd.to_datetime(Disney_total_gross["release_date"]).dt.year))
Disney_total_gross_with_year_column
```

 The data types are of type object. We need to clean data by assigning the correct type (float) to total_gross
  after removing the $ sign and commas, so as to make plotting and calculations possible.

```python
# The data types are of type object. We need to clean data by assigning the correct type (float) to total_gross
# after removing the $ sign and commas, so as to make plotting and calculations possible

Disney_total_gross_with_year_column = (Disney_total_gross_with_year_column
    .assign(total_gross_number= Disney_total_gross_with_year_column['total_gross']
        .replace('[$|,]','',regex=True).astype(float)))
Disney_total_gross_with_year_column
```

 Sort the data based on the total gross and drop the rows where total gross is not available

```python
# sort the data based on the total gross and drop the rows where total gross is not available
Disney_total_gross_with_year_column.sort_values("total_gross_number").dropna()
```

### Question 7. What is the Total Gross amount Disney made from movies in each year from 1937-2016?
###    (Why: To get an idea of the ups and downs of animation movie industry )

```python
#group the data frame by year and aggregrate sum of total_gross expressed as number
Disney_total_gross_grouped_by_year= (Disney_total_gross_with_year_column.groupby('year', as_index= False)
    .agg({"total_gross_number" :'sum'}))
Disney_total_gross_grouped_by_year
```

### Question 8. Plot the Total gross amount against year to get a visual idea of the question above.
### (Why: to visualize the troughs and peaks of gross collection)

```python
#plot the total gross Disney made in every year in the data frame
chart_movienumber_vs_gross = (alt.Chart(Disney_total_gross_grouped_by_year, width =800, height = 300).mark_bar()
    .encode(x="year", y= "total_gross_number")
    .properties(title =" Fig 2.Total Gross amount Disney made in years from 1937-2016")
                             )
chart_movienumber_vs_gross
```

### Question 9. Plot the transition of maximum grossing amount and minimum grossing amount in years from 1937-2016
###   (Why: to see deeper into the years in terms of gross)

```python
#group the data frame by year and aggregrate max amount of total_gross expressed as number
Disney_max_gross_grouped_by_year= (Disney_total_gross_with_year_column.groupby('year', as_index= False)
    .agg({"total_gross_number" :'max'}))
Disney_max_gross_grouped_by_year

#group the data frame by year and aggregrate min amount of total_gross expressed as number
Disney_min_gross_grouped_by_year= (Disney_total_gross_with_year_column.groupby('year', as_index= False)
    .agg({"total_gross_number" :'min'}))
Disney_min_gross_grouped_by_year
```

```python
#clean the data omit values 0.0
Disney_max_gross_grouped_by_year = Disney_max_gross_grouped_by_year[Disney_max_gross_grouped_by_year['total_gross_number']!=0.0]
Disney_min_gross_grouped_by_year = Disney_min_gross_grouped_by_year[Disney_min_gross_grouped_by_year['total_gross_number']!=0.0]

#plot the transition of maximum grossing amount and minimum grossing amount in years from 1937-2016
line_max_gross = (alt.Chart(Disney_max_gross_grouped_by_year, width =800, height = 300).mark_line(color ="Green", opacity =0.5, size= 4)
    .encode(x="year", y = alt.Y("total_gross_number:Q" , title ="Max gross amount")).properties(title =" Fig 3. Max Grossing vs minimum Grossing amount Disney made in years from 1937-2016")
    )

line_min_gross = (alt.Chart(Disney_min_gross_grouped_by_year, width =800, height = 300).mark_line(color ="Red", opacity =0.4, size= 2)
    .encode(x="year", y = alt.Y("total_gross_number:Q", title = "Min gross amount")).properties(title =" Fig 3. Max Grossing vs minimum Grossing amount Disney made in years from 1937-2016")
                 )
line_max_gross + line_min_gross
```

The plot shows Maximum gross in years as a thick green line and minimum gross amount as thin red line

### Question 13. Whether there is any popular day in a week for a Disney movie release?

#### Load the data with release date as a datetime object

```python
#load the data frame and set release_date as a datetime object
disney_movies_with_release_day = pd.read_csv("data/disney_movies_total_gross.csv", parse_dates = ['release_date'])
```

#### Check the data types

```python
disney_movies_with_release_day.dtypes
disney_movies_with_release_day.info()
```

#### Assign a new column with the day of the week in it. The day of the week is extracted from the release_date which is of type datetime 

```python
#assign a new column with the day of the week in it. The day of the week is extracted from the release_date which is of type datetime 
disney_movies_with_release_day = disney_movies_with_release_day.assign(day_of_week = disney_movies_with_release_day['release_date'].dt.day_name()).dropna(subset = 'day_of_week')
disney_movies_with_release_day
```

```python
# group the entries by day day of the week, find the count and sort based on the count
(disney_movies_with_release_day.groupby("day_of_week").count()
    .reset_index().sort_values(by ="release_date", ascending = False))
```

```python
#find the day of the week with the most count
counts = disney_movies_with_release_day.groupby("day_of_week").size().sort_values(ascending = False)
counts.index[0]
```
